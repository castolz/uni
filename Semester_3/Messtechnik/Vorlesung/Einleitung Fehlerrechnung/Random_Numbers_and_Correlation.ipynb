Simulation of noise signals

The noise signals are represented by statistical sequences of numbers that follow a normal distribution.

```python
import numpy as np
import matplotlib.pyplot as plt

# number of elements
N = 10000

avg = 0
std = 1

# signal 1 (normally distributed, arithmetic mean = 0, standard deviation = 1)

signal_1 = np.random.normal(avg, std, N)

# signal 2 (normally distributed, arithmetic mean = 0, standard deviation = 1)
signal_2 = np.random.normal(avg, std, N)

# element-wise product
product = signal_1 * signal_2

# --- plotting ---

fig, axs = plt.subplots(3, 2, figsize=(12, 10))

# signal 1
axs[0, 0].plot(signal_1[:1000])
axs[0, 0].set_title("signal 1 (normally distributed, first 1000 elements only)")
axs[0, 1].hist(signal_1, bins=100, density=True)
axs[0, 1].set_title("histogram of signal 1")

# Folge 2
axs[1, 0].plot(signal_2[:1000])
axs[1, 0].set_title("signal 2 (normally distributed, first 1000 elements only)")
axs[1, 1].hist(signal_2, bins=100, density=True)
axs[1, 1].set_title("histogram of signal 2")

# Produkt
axs[2, 0].plot(product[:1000])
axs[2, 0].set_title("product (first 1000 elements only)")
axs[2, 1].hist(product, bins=100, density=True)
axs[2, 1].set_title("histogram of product")

plt.tight_layout()
plt.show()
```

Correlation between the signals

The correlation coefficient between signal_1 and signal_2 should be 0, indicating no correlation.

```python
# correlation coefficient
correlation = np.corrcoef(signal_1, signal_2)[0, 1]
print("correlation (signal_1, signal_2):", correlation)
```

The correlation coefficient between signal_1 and the product = signal_1 * signal_2 should be 0, indicating no correlation.

```python
correlation = np.corrcoef(signal_1, product)[0, 1]
print("correlation (signal_1, product):", correlation)
```

The correlation coefficient between signal_2 and the product = signal_1 * signal_2 should be 0, indicating no correlation.

```python
correlation =  np.corrcoef(signal_2, product)[0, 1]
print("correlation (signal_2, product):", correlation)
```

Of course, the generated sequences of numbers must be perfectly correlated with themselves:

```python
print("correlation (signal_1, signal_1):", np.corrcoef(signal_1, signal_1)[0, 1])
```

```python
print("correlation (signal_2, signal_2):", np.corrcoef(signal_2, signal_2)[0, 1])
```

The same applies to the product:

```python
print("correlation (product, product):", np.corrcoef(product, product)[0, 1])
```

The arithmetic mean values are now calculated. A value close to 0 is expected.

```python
mean = sum(signal_1)/N
print('arithmetic mean of signal_1 = ', mean)
```

```python
mean = sum(signal_2)/N
print('arithmetic mean of signal_2 = ', mean)
```

```python
mean = sum(product)/N
print('arithmetic mean of product = ', mean)
```

The root mean square values of signal_1 and signal_2 must be equal to their standard deviation:

```python
import math
rms = math.sqrt(sum(signal_1**2)/N)
print('signal_1 rms = ', rms)
```

```python
rms = math.sqrt(sum(signal_2**2)/N)
print('signal_2 rms = ', rms)
```

Calculating the rms value of the signal_1 + signal_2 by definition:

```python
rms = math.sqrt(sum((signal_1 + signal_2)**2)/N)
print('(signal_1 + signal_2) rms = ', rms)
```

Calculating the geometrical sum of signal_1 and signal_2 (neglecting the contribution of the sum over the product of signal_1 and  signal_2):

```python
rms = math.sqrt(sum(signal_1**2 + signal_2**2)/N)
print('(signal_1 + signal_2) rms = ', rms)
```

It is obvious that the contribution of the sum over the product of signal_1 and  signal_2 is negligible.
